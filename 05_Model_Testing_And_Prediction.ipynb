```python
# Import Libraries

import pandas as pd
import joblib
import os
```

```python
# Load Saved Model

model = joblib.load("../models/random_forest_model.pkl")

print("Model Loaded Successfully")
```

```python
# Load Final Feature Dataset
# Robust path lookup — works regardless of the exact dataset folder name/location

def find_file(filename, search_dirs=("../Dataset", "../dataset", "../data", "..", ".")):
    for d in search_dirs:
        candidate = os.path.join(d, filename)
        if os.path.exists(candidate):
            return candidate
    raise FileNotFoundError(f"Could not find '{filename}'. Please check the dataset folder path.")

features_path = find_file("final_features_v3_timewindowed.csv")
print("Using features file:", features_path)

features = pd.read_csv(features_path)

# NOTE: If you see duplicated columns ending in _x / _y below (e.g. max_login_drift_x,
# max_login_drift_y), it means the drift-feature cells in 03_Feature_Engineering.ipynb
# were run twice on top of an already-processed file. Fix: re-run 03_Feature_Engineering.ipynb
# starting from the ORIGINAL final_features.csv (not the v3 file) to regenerate a clean file.

features.head()
```

```python
# Select Sample User

sample = features.drop(columns=["user"]).iloc[[0]]

sample
```

```python
# Pass the exact features expected by the trained model
# (uses model.feature_names_in_ so it always matches whatever the model was trained on)

sample = features[model.feature_names_in_].iloc[[0]]

prediction = model.predict(sample)
```

```python
# Display Result

if prediction[0] == 1:
    print("Insider Threat Detected")
else:
    print("Normal User")
```
