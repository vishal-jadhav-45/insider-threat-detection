```python
# Import Libraries

import pandas as pd
import joblib
```

```python
# Load Saved Model

model = joblib.load("../models/random_forest_model.pkl")

print("Model Loaded Successfully")
```

```python
# Load Final Feature Dataset

features = pd.read_csv("../Dataset/final_features.csv")

features.head()
```

```python
# Select Sample User

sample = features.drop(columns=["user"]).iloc[[0]]

sample
```

```python
# Predict

prediction = model.predict(sample)

print(prediction)
```

```python
# Display Result

if prediction[0] == 1:
    print("⚠️ Insider Threat Detected")
else:
    print("✅ Normal User")
```
