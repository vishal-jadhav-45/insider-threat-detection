```python
import pandas as pd
```

```python
# Load All Required Datasets

logon_data = pd.read_csv("../dataset/logon.csv")
device_data = pd.read_csv("../dataset/device.csv")
email_data = pd.read_csv("../dataset/email.csv")
file_data = pd.read_csv("../dataset/file.csv")
psychometric_data = pd.read_csv("../dataset/psychometric.csv")
```

```python
# Convert Date Column into Datetime Format

logon_data['date'] = pd.to_datetime(logon_data['date'])
device_data['date'] = pd.to_datetime(device_data['date'])
email_data['date'] = pd.to_datetime(email_data['date'])
file_data['date'] = pd.to_datetime(file_data['date'])
```

```python
# Check Duplicate Records

print(logon_data.duplicated().sum())
print(device_data.duplicated().sum())
print(email_data.duplicated().sum())
print(file_data.duplicated().sum())
print(psychometric_data.duplicated().sum())
```

```python

# Remove Duplicate Records

logon_data = logon_data.drop_duplicates()
device_data = device_data.drop_duplicates()
email_data = email_data.drop_duplicates()
file_data = file_data.drop_duplicates()
psychometric_data = psychometric_data.drop_duplicates()
```

```python

# Verify Dataset Shape After Cleaning

print(logon_data.shape)
print(device_data.shape)
print(email_data.shape)
print(file_data.shape)
print(psychometric_data.shape)
```

```python
# Save Cleaned Data

logon_data.to_csv("../dataset/logon_cleaned.csv", index=False)
device_data.to_csv("../dataset/device_cleaned.csv", index=False)
email_data.to_csv("../dataset/email_cleaned.csv", index=False)
file_data.to_csv("../dataset/file_cleaned.csv", index=False)
psychometric_data.to_csv("../dataset/psychometric_cleaned.csv", index=False)
```
